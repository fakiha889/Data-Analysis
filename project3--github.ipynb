# **Task1 : Student Performance Dataset**
# **Decription**
**This dataset contains student study hours, attendance, assignment scores, exam scores, and results. We will clean missing values, calculate statistics, analyze student performance, create visualizations, and practice basic classification.**

```python
import pandas as pd
import matplotlib.pyplot as plt
# Data pre-processing
raw_laptop_sales = [12, None, 18, 15, None, 22, 20]
clean_laptop_sales = [val for val in raw_laptop_sales if val is not None]
print("Cleaned Laptop Sales List:", clean_laptop_sales)

# Statistics
total_units = sum(clean_laptop_sales)
average_units = total_units / len(clean_laptop_sales)
peak_sales = max(clean_laptop_sales)
print(f"Total Sold: {total_units} | Average: {average_units:.1f} | Peak: {peak_sales}")

# Programming
laptop_inventory = {"Model": "ThinkBook 15", "Stock": 35, "Price_USD": 750}
print(f"Product: {laptop_inventory['Model']} | In-Stock Units:{laptop_inventory['Stock']}")

# Mathmetics
sales_matrix = [[10, 15], [20, 25]] # 2x2 nested list matrix
doubled_matrix = [[element * 2 for element in row] for row in sales_matrix]
print("Original Matrix:", sales_matrix)
print("Doubled Matrix: ", doubled_matrix)

#Software engineering
def calculate_discounted_price(original_price, discount_percentage):
 """Calculate final price after deducting promotional discount."""
 return original_price * (1 - discount_percentage / 100)
final_price = calculate_discounted_price(800, 15)
print(f"Promotional Price: ${final_price:.2f}")

#Machine learning
def classify_customer_sentiment(rating):
 """Decision stump classifier: Threshold = 7.0"""
 return "Satisfied" if rating >= 7.0 else "At Risk"
scores = [8.5, 6.2, 9.0, 4.5]
predictions = [classify_customer_sentiment(score) for score in scores]
print("Predictions:", predictions)

#Data visualization
weeks = ["W1", "W2", "W3", "W4", "W5"]
plt.bar(weeks, clean_laptop_sales, color="navy")
plt.title("Weekly Laptop Sales Comparison")
plt.ylabel("Units Sold")
plt.show()

#Soft skill
story_summary = (
 f"Executive Briefing: Laptop sales grew from 12 units in Week 1 to a peak of 22 units in Week 4, "
 f"maintaining a solid average of {average_units:.1f} units weekly. "
 f"Inventory managers should reorder at least 20 units ahead of next week's anticipated demand."
)
print(story_summary)
```

# **Task2 : E-commerce Sales & Customer Behavior Dataset**
# **Description**
**This dataset contains product categories, prices, quantities, customer ratings, payment methods, marketing channels, and returns. We will clean data, analyze sales and revenue, compare product categories, visualize customer behavior, and practice predicting product returns.**

```python
import pandas as pd
import matplotlib.pyplot as plt
# data pre-processing
df = pd.read_csv("ecommerce1.csv")
# Check missing values
# Data before cleaning
print(df)
print(df.isnull().sum())
# Fill missing ratings with mode
df["Customer_Rating"] = df["Customer_Rating"].fillna(df["Customer_Rating"].mode()[0])
df["Marketing_Channel"] = df["Marketing_Channel"].fillna(df["Marketing_Channel"].mode()[0])
print("Missing values after cleaning:")
print(df.isnull().sum())

# Statistics
total_sales = sum(df["Total_Sales_USD"])
average_sales = total_sales/len(df["Total_Sales_USD"])
peak_sales = max(df["Total_Sales_USD"])
print("Total Sales : ",total_sales)
print("Average Sales : ",average_sales)
print("peak_sales : ",peak_sales)

#programming
Product = {
    "Product_Category":"Electronics",
    "Quantity":1,
    "Price_USD":299

}
print("product_cetagory :" ,Product["Product_Category"])
print("Quantity :",Product["Quantity"])
print("Price_USD :",Product["Price_USD"])

#mathematics
sales_matrix = [[60,70],
                [150,160]
                ]
double_sales = [[element * 2 for element in row] for row in sales_matrix]
print("Original Sales : ",sales_matrix)
print("double_sales : ",double_sales)

#software engineering
def calculate_price(price,discounted_price):
  return price * (1-discounted_price/100)
print(calculate_price(299,10))

#machine learning
def classify_customer_sentiment(rating):
  return "Not Good" if rating<4.0 else "Good Rating"
scores = [4.0,5.0,3.0,2.0]
prediction = [classify_customer_sentiment(score) for score in scores]
print("Rating is : ",prediction)

#Data visualization
import matplotlib.pyplot as plt
category_sales = df.groupby("Product_Category")["Total_Sales_USD"].sum()
plt.bar(category_sales.index, category_sales.values)
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (USD)")
plt.tight_layout()
plt.show()

#soft skill
story_summary = (
    "Total sales were 3124 with highest sales product was electronics"
    "The business should analyze customer preferences and "
    "promote other product categories to improve sales."
)
print("story_summary : ",story_summary)




```
